```python
import numpy as np
import pandas as pd
from scipy.integrate import trapezoid

from IPython.display import display

import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['grid.linestyle'] = "--"
plt.rcParams['grid.alpha'] = 0.5
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.spines.left'] = False
plt.rcParams['axes.spines.bottom'] = False
plt.rcParams['axes.labelsize'] = 10
```

```python
T = 1
N = 5000 # Numero de simulaciones/trayectorias
M = 1000 # Numero de pasos
s = np.linspace(0, T, M+1)

def mov_browniano(n, T, time_output = True):
    dt = np.sqrt(T/n)
    w = np.zeros(n + 1)
    t = np.linspace(0, T, n+1)
    w[0] = 0
    for i in range(1, n+1):
        z = np.random.normal(0,1)
        w[i] = w[i-1] + dt*z
    if time_output: return w, t
    else: return w

W = np.zeros((N,M+1))
W[0], t = mov_browniano(M, T)
for i in range(1,N):
    W[i] = mov_browniano(M, T, time_output=False)

integral_W = np.array([trapezoid(W[i]**2, s) for i in range(N)])
```

```python
mean_integral = np.mean(integral_W)
print(f"Valor analítico de la integral (t^2/2): {1/2:.4f}")
print(f"Valor aproximado: {mean_integral:.4f}\n")

var_integral = np.var(integral_W)
print(f"Valor analítico de la varianza de la integral: {1/3:.4f}")
print(f"Varianza de la integral: {var_integral:.4f}")
```
